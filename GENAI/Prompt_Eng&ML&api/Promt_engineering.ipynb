```python
from langchain_google_genai import ChatGoogleGenerativeAI
```

```python
model = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash-lite",
    temperature=0.3
    )
model
```

```python
reviews = [
    "The product quality is amazing and delivery was very fast.",

    "The packaging was okay but the product stopped working after two days.",

    "Average product. Nothing special but not bad either.",

    "Excellent battery backup and smooth performance.",

    "Terrible customer support and poor build quality.",

    "The camera quality is fantastic but the battery drains quickly.",

    "Very comfortable shoes and worth the money.",

    "The food was cold and the service was disappointing.",

    "Nice design but performance is average.",

    "Absolutely loved the laptop. Super fast and lightweight."
]
```

```python
few_shot_prompt = """
You are a sentiment analysis expert.

Your task is to analyze customer reviews
using Chain-of-Thought reasoning.

Follow these steps carefully:

1. Identify all positive sentiment phrases.
2. Identify all negative sentiment phrases.
3. Check whether the review contains mixed sentiment.
4. Decide the final sentiment label:
   - Positive
   - Neutral
   - Negative
5. Explain the reasoning step by step.

--------------------------------------------------

Example 1:

Review:
"The mobile phone is excellent and battery backup is amazing."

Reasoning:
- Positive phrases:
  - "excellent"
  - "battery backup is amazing"

- Negative phrases:
  - None

- Mixed sentiment:
  - No

- Final Decision:
  The review contains strong positive opinions only.

Sentiment: Positive

--------------------------------------------------

Example 2:

Review:
"The food taste was okay but service was very slow."

Reasoning:
- Positive phrases:
  - "taste was okay"

- Negative phrases:
  - "service was very slow"

- Mixed sentiment:
  - Yes

- Final Decision:
  The review contains both positive and negative opinions.
  Therefore, the sentiment is neutral.

Sentiment: Neutral

--------------------------------------------------

Example 3:

Review:
"The laptop stopped working in one week and customer support was terrible."

Reasoning:
- Positive phrases:
  - None

- Negative phrases:
  - "stopped working"
  - "customer support was terrible"

- Mixed sentiment:
  - No

- Final Decision:
  The review expresses strong dissatisfaction.

Sentiment: Negative

--------------------------------------------------

Now analyze the following review carefully.

Review:
"""
```

```python
for i, review in enumerate(reviews, start=1):

    print("\n")
    print("=" * 70)
    print(f"REVIEW {i}")
    print("=" * 70)

    print("\nCustomer Review:")
    print(review)

    # Create final prompt
    final_prompt = few_shot_prompt + f'"{review}"'

    # Generate Response
    response = model.invoke(final_prompt)

    print("\nLLM OUTPUT:\n")

    print(response.content)

    print("\n")
```
